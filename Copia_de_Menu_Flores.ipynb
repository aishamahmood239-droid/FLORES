## Datos: 12 flores, su tipo de planta, significado, color y temporada

# Reto: Opciones opcionales — Menú de Flores 🌸

Script con un menú sobre flores. Reúne los tres conceptos que pide el reto:

- **Funciones**: cada opción del menú es una función distinta.
- **Condicionales**: el menú decide qué función llamar según la opción elegida (`if/elif/else`).
- **Bucles**: uno para que el menú se repita hasta elegir "Salir", y otro para validar
  que la flor que escribe el usuario existe.

Estructura pensada para poder explicarla paso a paso: primero los datos, luego cada
función por separado, y al final el menú que las conecta.

```python
FLORES = {
    "Tulipan":     {"tipo": "Bulbosa", "color": "Variado", "temporada": "Primavera",
                     "significado": "Amor perfecto y declaración."},
    "Lirio":       {"tipo": "Bulbosa", "color": "Blanco", "temporada": "Primavera",
                     "significado": "Pureza y renovación."},
    "Peonia":      {"tipo": "Perenne", "color": "Rosa", "temporada": "Primavera",
                     "significado": "Prosperidad y buena fortuna."},

    "Rosa":        {"tipo": "Arbusto", "color": "Rojo", "temporada": "Verano",
                     "significado": "Amor y pasión."},
    "Girasol":     {"tipo": "Anual", "color": "Amarillo", "temporada": "Verano",
                     "significado": "Adoración y lealtad."},
    "Jazmin":      {"tipo": "Trepadora", "color": "Blanco", "temporada": "Verano",
                     "significado": "Amabilidad y sensualidad."},

    "Dalia":       {"tipo": "Bulbosa", "color": "Variado", "temporada": "Otoño",
                     "significado": "Compromiso y cambio."},
    "Orquidea":    {"tipo": "Perenne", "color": "Variado", "temporada": "Otoño",
                     "significado": "Belleza exótica y lujo."},
    "Crisantemo":  {"tipo": "Perenne", "color": "Variado", "temporada": "Otoño",
                     "significado": "Honestidad y lealtad."},

    "Camelia":     {"tipo": "Arbusto", "color": "Rojo/Rosa", "temporada": "Invierno",
                     "significado": "Admiración y perfección."},
    "Ciclamen":    {"tipo": "Bulbosa", "color": "Rosa/Blanco", "temporada": "Invierno",
                     "significado": "Timidez y despedida."},
    "Amarilis":    {"tipo": "Bulbosa", "color": "Rojo", "temporada": "Invierno",
                     "significado": "Determinación y belleza radiante."},
}

# Orden cíclico de las estaciones, para comparar temporadas
ESTACIONES = ["Primavera", "Verano", "Otoño", "Invierno"]
```

## Función auxiliar — validar la entrada del usuario

```python
def pedir_flor(mensaje="Escribe el nombre de una flor: "):
    """
    Pide una flor por teclado y no continúa hasta que sea una válida
    (bucle de validación, ignora mayúsculas/minúsculas y acentos simples).
    """
    while True:
        flor = input(mensaje).strip().capitalize()
        if flor in FLORES:
            return flor
        print("   Esa flor no está en la lista. Prueba con, por ejemplo: Rosa, Tulipan, Girasol...")
```

## Funciones — una por cada opción del menú

```python
def listar_flores():
    """Opción 1: muestra las 12 flores con su temporada."""
    print("\n📋 Flores disponibles:")
    for flor, datos in FLORES.items():
        print(f"  - {flor}: florece en {datos['temporada']}")


def consultar_significado():
    """Opción 2: muestra el significado de una flor."""
    flor = pedir_flor()
    significado = FLORES[flor]["significado"]
    print(f"\n💌 {flor}: {significado}")


def consultar_tipo():
    """Opción 3: muestra el tipo de planta y el color habitual de una flor."""
    flor = pedir_flor()
    tipo = FLORES[flor]["tipo"]
    color = FLORES[flor]["color"]
    print(f"\n🌱 {flor} es una planta de tipo {tipo}, normalmente de color {color}.")


def comparar_temporada():
    """Opción 4: compara en qué temporada florece cada una de dos flores."""
    print("Vamos a comparar la temporada de dos flores:")
    flor1 = pedir_flor("Primera flor: ")
    flor2 = pedir_flor("Segunda flor: ")
    temporada1 = FLORES[flor1]["temporada"]
    temporada2 = FLORES[flor2]["temporada"]

    indice1 = ESTACIONES.index(temporada1)
    indice2 = ESTACIONES.index(temporada2)
    distancia = min(abs(indice1 - indice2), 4 - abs(indice1 - indice2))

    if distancia == 0:
        resultado = "🌼 Florecen en la misma temporada."
    elif distancia == 1:
        resultado = "🌿 Sus temporadas se solapan un poco."
    else:
        resultado = "❄️ Florecen en temporadas opuestas."

    print(f"\n{flor1} ({temporada1}) + {flor2} ({temporada2}): {resultado}")
```

## Programa principal — el menú

Un `while` mantiene el menú activo hasta elegir "Salir", y un `if/elif/else`
decide qué función llamar según la opción escrita.

```python
def mostrar_menu():
    print("\n🌸 --- Explorador de Flores --- 🌸")
    print("1. Ver todas las flores y su temporada")
    print("2. Consultar el significado de una flor")
    print("3. Consultar el tipo de planta y color de una flor")
    print("4. Comparar la temporada de dos flores")
    print("5. Salir")
    return input("Elige una opción (1-5): ").strip()


def main():
    opcion = ""
    while opcion != "5":
        opcion = mostrar_menu()

        if opcion == "1":
            listar_flores()
        elif opcion == "2":
            consultar_significado()
        elif opcion == "3":
            consultar_tipo()
        elif opcion == "4":
            comparar_temporada()
        elif opcion == "5":
            print("\n👋 ¡Hasta la próxima!")
        else:
            print("\n⚠️  Opción no válida, elige un número del 1 al 5.")
```

## Ejecutar el menú

```python
main()
```
